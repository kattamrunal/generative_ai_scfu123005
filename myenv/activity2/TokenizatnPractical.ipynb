```python
from transformers import AutoTokenizer
#AutoTokenizer is a class from the Hugging Face Transformers library that provides a convenient way to tokenize text for various pre-trained models. It automatically selects the appropriate tokenizer based on the specified model name or path.
```

```python
tokenizer=AutoTokenizer.from_pretrained("bert-base-uncased")
# creating object named "tokenizer" of class AutoTokenizer and loading the pre-trained tokenizer for the BERT model with the "bert-base-uncased" configuration. This tokenizer will be used to convert text into token IDs that can be fed into the BERT model for various NLP tasks.
```

```python
text = "This is Katta Mrunal"
```

```python
#Step 1 - tokenize the input
tokenizer.tokenize(text)
```

```python
#convert token into token id's
tokenizer.encode(text)
```
