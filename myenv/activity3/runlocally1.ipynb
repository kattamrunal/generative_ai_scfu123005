```python
# go in website -> docs -> copy code and run it in your local machine
import os
from dotenv import load_dotenv

from groq import Groq

load_dotenv()

groq_api = os.environ.get("GROQ_API_KEY")

client = Groq(
    # api_key=groq_api
)

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user", #we use model as user 
            "content": "Explain the importance of fast language models", # this is the prompt we are giving to the model
        }
    ],
    model="llama-3.3-70b-versatile",
)

print(chat_completion.choices[0].message.content)
#we will get the answer below
```
